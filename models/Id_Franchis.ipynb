```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data")

franchise_mapping = {
    # Charlotte
    "CHH": "CHA",
    "CHO": "CHA",

    # Denver
    "DNR": "DEN",
    "DNA": "DEN",
    "DEN": "DEN",

    # Brooklyn / New Jersey
    "NJN": "BRK",
    "BRK": "BRK",
}

files_to_rewrite = [
    "Team_Stats_Per_100_Poss.csv",
    "Opponent_Stats_Per_100_Poss.csv",
    "Team_Summaries.csv",
    "Team_Abbrev.csv",
]

for filename in files_to_rewrite:

    input_path = DATA_DIR / filename

    # Vérifier que le fichier existe
    if not input_path.exists():
        print(f"Fichier introuvable : {input_path.resolve()}")
        continue

    # Nouveau fichier
    output_path = DATA_DIR / (
        input_path.stem + "_franchise" + input_path.suffix
    )

    # Lecture
    df = pd.read_csv(input_path)

    # Vérifier la colonne
    if "abbreviation" not in df.columns:
        print(f"⚠{filename} : pas de colonne 'abbreviation'")
        continue

    # Création du franchise_id
    df["franchise_id"] = (
        df["abbreviation"]
        .map(franchise_mapping)
        .fillna(df["abbreviation"])
    )

    # Écriture avec pandas
    df.to_csv(output_path, index=False)

    print(f"{filename}")
    print(f"   → {output_path.resolve()}")
```
